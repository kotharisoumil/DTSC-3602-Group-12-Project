```python
import geopandas as gpd
from shapely.geometry import Point
```

```python
crime_gdf = gpd.GeoDataFrame(
    crime,
    geometry=gpd.points_from_xy(crime["LON"], crime["LAT"]),
    crs="EPSG:4326"
)

crime_gdf.head()
```

```python
tracts = tracts.to_crs("EPSG:4326")
tracts.head()
```

```python
tracts_for_join = tracts[["GEOID", "geometry"]]

crime_with_tract = gpd.sjoin(
    crime_gdf,
    tracts_for_join,
    how="left",
    predicate="within"
)

crime_with_tract.head()
```

```python
crime_with_tract["GEOID"].isna().mean()
```

```python
crime_with_tract[['DR_NO','LAT','LON','GEOID']].head()
```

```python
# Add prefix to all ACS columns so teammates know they came from ACS
acs = acs.add_prefix("acs_")

# Fix the geoid name back to 'geoid' for merging
acs = acs.rename(columns={"acs_geoid": "geoid"})

```

```python
acs = acs.rename(columns={
    "acs_b01003_001_moe": "acs_total_pop_moe",
    "acs_b19013_001_moe": "acs_median_income_moe"
})

acs.head()
```

```python
# Make sure both GEOID keys are strings with leading zeros to length 11
crime_with_tract["geoid"] = crime_with_tract["GEOID"].astype(str).str.zfill(11)

# (After you've reloaded/cleaned ACS and set prefix/renames)
acs["geoid"] = acs["geoid"].astype(str).str.zfill(11)
```

```python
# Standardize geoid columns as strings
crime_with_tract["geoid"] = crime_with_tract["GEOID"].astype(str).str.strip()
acs["geoid"] = acs["geoid"].astype(str).str.strip()

# Merge crime+tract with ACS contextual features
merged = crime_with_tract.merge(
    acs,
    on="geoid",
    how="left"
)

merged.head()
merged.shape
```

```python
merged_clean = merged.drop(columns=["geometry", "index_right"], errors="ignore")

merged_clean.shape
merged_clean.columns

```

```python
output_path = "/Users/veer.pth/Desktop/Sem 5/Data Science 3/Data Science 2/context_enriched_dataset.csv"

merged_clean.to_csv(output_path, index=False)

```
